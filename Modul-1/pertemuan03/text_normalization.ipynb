```python
def case_folding(text):
    return str(text).lower()

# Contoh
text = "Saya Belajar NLP"
print(case_folding(text))
# Output: saya belajar nlp
```

```python
def remove_url(text):
    return re.sub(r"http\S+|www\S+", "", text)

# Contoh
text = "Kunjungi https://example.com sekarang"
print(remove_url(text))
# Output: Kunjungi  sekarang
```

```python
def remove_mention(text):
    return re.sub(r"@\w+", "", text)

# Contoh
text = "@user1 produk ini bagus"
print(remove_mention(text))
```

```python
import re
import pandas as pd

def remove_hashtag(text):
    if pd.isna(text):
        return text
    return re.sub(r"#\w+", "", text)

df["review_bersih"] = df["review"].apply(remove_hashtag)

display(df[["review", "review_bersih"]].head())
```

```python
def remove_number(text):
    return re.sub(r"\d+", "", text)
```

```python
def remove_punctuation(text):
    return text.translate(
        str.maketrans("", "", string.punctuation)
    )
```

```python
def clean_whitespace(text):
    return re.sub(r"\s+", " ", text).strip()
```

```python
def clean_text(text):
    text = str(text)
    text = text.lower()           # case folding
    text = remove_url(text)       # hapus URL
    text = remove_mention(text)   # hapus mention
    text = remove_hashtag(text)   # hapus hashtag
    text = remove_number(text)    # hapus angka
    text = remove_punctuation(text)  # hapus punctuation
    text = clean_whitespace(text) # bersihkan whitespace
    return text
```

```python
import re
import string
import pandas as pd

text_column = "review"

def clean_text(text):
    if pd.isna(text):
        return text

    text = re.sub(r"#\w+", "", text)  # Hapus hashtag beserta katanya
    text = text.translate(str.maketrans("", "", string.punctuation))
    text = re.sub(r"\s+", " ", text).strip()  # Rapikan spasi
    return text

df["clean_text"] = df[text_column].apply(clean_text)

display(df[[text_column, "clean_text"]].head(10))
```

```python
stop_words = set(stopwords.words("indonesian"))

def remove_stopwords(tokens):
    return [
        token for token in tokens
        if token not in stop_words
    ]

# Contoh
tokens = ["saya", "sedang", "belajar", "nlp", "yang", "menarik"]
print(remove_stopwords(tokens))
```

```python
factory = StemmerFactory()
stemmer = factory.create_stemmer()

def stem_tokens(tokens):
    return [stemmer.stem(token) for token in tokens]

# Contoh
text = "Mahasiswa sedang mempelajari pemrograman."
print(stemmer.stem(text))
```

```python
import re

def remove_url(text):
    return re.sub(r"https?://\S+|www\.\S+", "", str(text))

def remove_mention(text):
    return re.sub(r"@\w+", "", str(text))
```

```python

def preprocessing_pipeline(text):
    # 1. Cleaning
    text = clean_text(text)

    # 2. Tokenization
    tokens = word_tokenize(text)

    # 3. Stopword removal
    tokens = remove_stopwords(tokens)

    # 4. Stemming
    tokens = stem_tokens(tokens)

    return tokens

# Menerapkan ke dataset
df["processed_tokens"] = df[text_column].apply(preprocessing_pipeline)
```

```python
df["processed_text"] = df["processed_tokens"].apply(
    lambda tokens: " ".join(tokens)
)
```

```python
import re
import pandas as pd
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory

# Baca dataset (sesuaikan path jika lokasi notebook berbeda)
df = pd.read_csv("../dataset/master.csv")
text_column = "review"

# Siapkan stemmer dan daftar stopword Bahasa Indonesia
stemmer = StemmerFactory().create_stemmer()
stopwords = set(StopWordRemoverFactory().get_stop_words())

# Ambil 20 ulasan yang tidak kosong
raw_texts = df[text_column].dropna().astype(str).head(20)

if len(raw_texts) < 20:
    raise ValueError(f"Dataset hanya memiliki {len(raw_texts)} ulasan yang tidak kosong.")

hasil = []

for raw_text in raw_texts:
    # Lowercase
    lowercase = raw_text.lower()

    # Cleaning: hapus URL, mention, hashtag beserta katanya, tanda baca, dan spasi berlebih
    cleaned = re.sub(r"https?://\S+|www\.\S+", " ", lowercase)
    cleaned = re.sub(r"@\w+|#\w+", " ", cleaned)
    cleaned = re.sub(r"[^\w\s]", " ", cleaned)
    cleaned = re.sub(r"\s+", " ", cleaned).strip()

    # Tokenisasi
    tokens = re.findall(r"\b\w+\b", cleaned)

    # Hapus stopword
    tokens_without_stopwords = [
        token for token in tokens if token not in stopwords
    ]

    # Stemming
    stemmed_text = stemmer.stem(" ".join(tokens_without_stopwords))

    hasil.append({
        "Raw Text": raw_text,
        "Setelah Lowercase": lowercase,
        "Setelah Cleaning": cleaned,
        "Setelah Tokenisasi": tokens,
        "Setelah Stopword": tokens_without_stopwords,
        "Setelah Stemming": stemmed_text,
    })

tabel_eksperimen = pd.DataFrame(hasil)
display(tabel_eksperimen)
```
