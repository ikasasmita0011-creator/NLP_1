```python
import pandas as pd
import nltk
import spacy
from nltk.tokenize import word_tokenize

# Unduh data yang dibutuhkan NLTK (cukup sekali)
nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)

# spaCy multilingual tokenizer tanpa model terlatih
nlp = spacy.blank("xx")

texts = [
    "Saya belajar NLP.",
    "Bagus banget!!!",
    "Harga Rp50.000",
]

hasil = []

for teks in texts:
    hasil.append({
        "Teks": teks,
        "NLTK word_tokenize": word_tokenize(teks, language="english"),
        "split()": teks.split(),
        "spaCy": [token.text for token in nlp(teks)],
    })

tabel_perbandingan = pd.DataFrame(hasil)
display(tabel_perbandingan)
```
