```python
import pandas as pd

def analyze_pos(text, nlp_model):
    doc = nlp_model(text)
    results = []
    for token in doc:
        results.append({
            "Token": token.text,
            "POS": token.pos_,
            "POS Tag": token.tag_,
            "Lemma": token.lemma_,
            "Dependency": token.dep_,
            "Head": token.head.text
        })
    return pd.DataFrame(results)

df_pos = analyze_pos("Mahasiswa belajar NLP.", nlp)
print(df_pos.to_string(index=False))
```

```python
sentences = [
    "Mahasiswa belajar NLP.",
    "Dosen mengajar Natural Language Processing.",
    "Mahasiswa membaca buku.",
    "Sistem memproses data teks.",
    "Model bahasa memprediksi kata berikutnya."
]

for sentence in sentences:
    print(f"\nKalimat: {sentence}")
    df_result = analyze_pos(sentence, nlp)
    print(df_result[["Token", "POS", "Dependency", "Head"]].to_string(index=False))
    print()
```

```python
import pandas as pd

df = pd.read_csv("../dataset/dataset_preprocessed.csv")
text_column = "processed_text"

# Ambil baris yang memiliki teks
valid_rows = df[
    df[text_column].notna()
    & df[text_column].astype(str).str.strip().ne("")
]

if valid_rows.empty:
    raise ValueError(f"Kolom '{text_column}' tidak berisi teks untuk dianalisis.")

# Ambil maksimal 10 sampel
sample = valid_rows.sample(n=min(10, len(valid_rows)), random_state=42)

for _, row in sample.iterrows():
    text = str(row[text_column])[:200]  # batasi panjang teks
    doc = nlp(text)

    pos_counts = {}
    for token in doc:
        pos = token.pos_
        pos_counts[pos] = pos_counts.get(pos, 0) + 1

    print(f"Teks: {text[:80]}{'...' if len(text) > 80 else ''}")
    print(f"POS Distribution: {pos_counts}")
    print()
```

```python
import pandas as pd
from spacy import displacy
from IPython.display import display

df = pd.read_csv("../dataset/dataset_preprocessed.csv")
text_column = "processed_text"

# Validasi kolom dan pipeline
df.columns = df.columns.str.strip()

if text_column not in df.columns:
    raise KeyError(f"Kolom '{text_column}' tidak ditemukan. Kolom tersedia: {df.columns.tolist()}")

if "parser" not in nlp.pipe_names:
    raise ValueError(
        f"Pipeline tidak memiliki dependency parser. Komponen aktif: {nlp.pipe_names}"
    )

# Siapkan teks yang tidak kosong
texts = (
    df[text_column]
    .dropna()
    .astype(str)
    .str.strip()
)
texts = texts[texts != ""]

# Ambil 10 kalimat dari dataset
sentences = []

for doc in nlp.pipe(texts.tolist(), batch_size=16):
    for sentence in doc.sents:
        if sentence.text.strip():
            sentences.append(sentence.as_doc())
        if len(sentences) == 10:
            break
    if len(sentences) == 10:
        break

if len(sentences) < 10:
    raise ValueError(
        f"Hanya ditemukan {len(sentences)} kalimat. "
        "Dataset perlu memiliki setidaknya 10 kalimat yang tidak kosong."
    )

# Buat tabel hasil analisis
results = []

for number, doc in enumerate(sentences, start=1):
    tokens = [token.text for token in doc]

    pos_tags = [
        f"{token.text}: {token.pos_}"
        for token in doc
    ]

    dependencies = [
        f"{token.text} → {token.head.text} ({token.dep_})"
        for token in doc
    ]

    # Pemeriksaan otomatis sederhana, bukan pengganti pemeriksaan manual
    issues = []
    uncertain_pos = [token.text for token in doc if token.pos_ in ("", "X")]
    uncertain_dep = [token.text for token in doc if token.dep_ in ("", "dep")]
    root_count = sum(token.dep_ == "ROOT" for token in doc)

    if uncertain_pos:
        issues.append(f"POS perlu diperiksa: {', '.join(uncertain_pos)}")
    if uncertain_dep:
        issues.append(f"Dependency perlu diperiksa: {', '.join(uncertain_dep)}")
    if root_count != 1:
        issues.append(f"Jumlah ROOT tidak biasa: {root_count}")

    error_note = "; ".join(issues) if issues else "Tidak ada anomali otomatis; cek manual"

    results.append({
        "No.": number,
        "Kalimat": doc.text,
        "Token": " | ".join(tokens),
        "POS": " | ".join(pos_tags),
        "Dependency": " | ".join(dependencies),
        "Kesalahan?": error_note,
    })

table = pd.DataFrame(results)
display(table)

# Tampilkan dependency tree untuk setiap kalimat
for number, doc in enumerate(sentences, start=1):
    print(f"Dependency tree kalimat {number}: {doc.text}")
    display(displacy.render(doc, style="dep", jupyter=True))
```
